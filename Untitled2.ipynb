```python
# Cell 3 — inspect files & class counts
import os
# possible dataset folder names: 'train' (as in dataset)
train_dir = '/content/orchid_dataset/train'
if not os.path.exists(train_dir):
    # try to find directory
    for f in os.listdir('/content/orchid_dataset'):
        if os.path.isdir(os.path.join('/content/orchid_dataset', f)) and 'train' in f.lower():
            train_dir = os.path.join('/content/orchid_dataset', f)
            break

print("Using train dir:", train_dir)
classes = [d for d in os.listdir(train_dir) if os.path.isdir(os.path.join(train_dir, d))]
print("Number of classes found:", len(classes))
# print counts per class (first 20)
counts = {c: len([f for f in os.listdir(os.path.join(train_dir,c)) if f.lower().endswith(('jpg','jpeg','png'))]) for c in classes}
print("Example class counts (first 20):")
for i,(k,v) in enumerate(sorted(counts.items(), key=lambda x:-x[1])[:20]):
    print(f"{i+1:02d}. {k}: {v}")
```

```python
# Cell 4 — prepare datasets using image_dataset_from_directory
import tensorflow as tf
from tensorflow.keras.preprocessing import image_dataset_from_directory

IMG_SIZE = (224, 224)   # VGG16 input
BATCH_SIZE = 32
SEED = 123

# If there is a separate test folder, adapt — here we assume only a train folder so we split it
train_ds = image_dataset_from_directory(
    train_dir,
    validation_split=0.2,
    subset="training",
    seed=SEED,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE
)

val_ds = image_dataset_from_directory(
    train_dir,
    validation_split=0.2,
    subset="validation",
    seed=SEED,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE
)

class_names = train_ds.class_names
num_classes = len(class_names)
print("Num classes:", num_classes)
print(class_names[:10])
# performance: prefetch
AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.prefetch(buffer_size=AUTOTUNE)
val_ds = val_ds.prefetch(buffer_size=AUTOTUNE)
```

```python
# Cell 5 — augmentation + preprocess (use VGG16 preprocess_input)
from tensorflow.keras import layers

data_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1),
])

# We'll use tf.keras.applications.vgg16.preprocess_input during model building/training mapping
from tensorflow.keras.applications.vgg16 import preprocess_input

def preprocess_dataset(ds):
    # ds yields (images, labels)
    return ds.map(lambda x,y: (preprocess_input(x), y), num_parallel_calls=AUTOTUNE)

train_ds = train_ds.map(lambda x,y: (data_augmentation(x, training=True), y), num_parallel_calls=AUTOTUNE)
train_ds = preprocess_dataset(train_ds)
val_ds = preprocess_dataset(val_ds)
```

```python
# Cell 6 — build model using VGG16 as base (transfer learning)
from tensorflow.keras.applications import VGG16
from tensorflow.keras import Model, Input
from tensorflow.keras import layers, optimizers

base_model = VGG16(include_top=False, weights='imagenet', input_shape=(224,224,3))
base_model.trainable = False   # freeze for initial training

inputs = Input(shape=(224,224,3))
x = base_model(inputs, training=False)   # feature extractor
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dense(512, activation='relu')(x)
x = layers.Dropout(0.5)(x)
outputs = layers.Dense(num_classes, activation='softmax')(x)
model = Model(inputs, outputs)

model.compile(optimizer=optimizers.Adam(learning_rate=1e-4),
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.summary()
```

```python
# Cell 7 — callbacks & compute class weights if you want to mitigate imbalance
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping, ReduceLROnPlateau

# Save best model to Drive (optional — mount Drive if you want persistent storage)
# from google.colab import drive
# drive.mount('/content/drive')

checkpoint_path = "/content/orchid_vgg16_best.h5"
checkpoint = ModelCheckpoint(checkpoint_path, monitor='val_accuracy', save_best_only=True)

early = EarlyStopping(monitor='val_accuracy', patience=6, restore_best_weights=True)
reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3, min_lr=1e-7)

callbacks = [checkpoint, early, reduce_lr]

# Compute class weights (optional)
import numpy as np
from collections import Counter
all_labels = []
for images, labels in train_ds.unbatch().take(1000000):  # gather labels (should fit)
    all_labels.append(int(labels.numpy()))
label_counts = Counter(all_labels)
print("Label counts (sample):", list(label_counts.items())[:10])
classes_arr = np.array([label_counts[i] for i in range(num_classes)])
from sklearn.utils import class_weight
class_weights = class_weight.compute_class_weight(class_weight='balanced',
                                                  classes=np.arange(num_classes),
                                                  y=np.array(all_labels))
class_weights = dict(enumerate(class_weights))
print("Class weights sample:", list(class_weights.items())[:5])
```

```python
# Cell 8 — initial training (frozen base)
EPOCHS = 20
history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS,
    callbacks=callbacks,
    class_weight=class_weights  # optional; remove if you don't want to use it
)
```

```python
# Cell 9 — fine-tuning: unfreeze last conv blocks of VGG16
base_model.trainable = True

# Freeze all layers except the last N layers (try N = 4-8)
fine_tune_at = len(base_model.layers) - 8  # tweak this
for layer in base_model.layers[:fine_tune_at]:
    layer.trainable = False
for layer in base_model.layers[fine_tune_at:]:
    layer.trainable = True

# recompile with smaller LR
model.compile(optimizer=optimizers.Adam(learning_rate=1e-5),
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

fine_epochs = 10
total_epochs = EPOCHS + fine_epochs

history_fine = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=total_epochs,
    initial_epoch=history.epoch[-1] if hasattr(history, 'epoch') else 0,
    callbacks=callbacks,
    class_weight=class_weights
)
```

```python
# Cell 10 — training curves
import matplotlib.pyplot as plt

def plot_history(h1, h2=None):
    hist = {}
    for k,v in h1.history.items(): hist[k] = v
    if h2:
        for k,v in h2.history.items():
            hist[k].extend(v)
    epochs = range(1, len(hist['loss'])+1)
    plt.figure(figsize=(14,5))
    plt.subplot(1,2,1)
    plt.plot(epochs, hist['loss'], label='train_loss')
    plt.plot(epochs, hist['val_loss'], label='val_loss')
    plt.xlabel('Epoch'); plt.ylabel('Loss'); plt.legend()
    plt.subplot(1,2,2)
    plt.plot(epochs, hist['accuracy'], label='train_acc')
    plt.plot(epochs, hist['val_accuracy'], label='val_acc')
    plt.xlabel('Epoch'); plt.ylabel('Accuracy'); plt.legend()
    plt.show()

plot_history(history, history_fine if 'history_fine' in globals() else None)
```

```python
# Cell 11 — evaluation + confusion matrix
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt
import itertools

# get true labels & preds on validation set
y_true = np.array([])
y_pred = np.array([])
y_prob = []

for imgs, labels in val_ds.unbatch().batch(64):
    preds = model.predict(imgs)
    pred_labels = np.argmax(preds, axis=1)
    y_true = np.concatenate([y_true, labels.numpy()])
    y_pred = np.concatenate([y_pred, pred_labels])
    y_prob.extend(preds)

print("Classification report:")
print(classification_report(y_true, y_pred, target_names=class_names))

cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(10,8))
plt.imshow(cm, interpolation='nearest', cmap=plt.cm.Blues)
plt.title('Confusion matrix (validation)')
plt.colorbar()
tick_marks = np.arange(len(class_names))
plt.xticks(tick_marks, class_names, rotation=90, fontsize=6)
plt.yticks(tick_marks, class_names, fontsize=6)
plt.tight_layout()
plt.ylabel('True label')
plt.xlabel('Predicted label')
plt.show()
```

```python
# Cell 13 — inference for a user-uploaded image
from google.colab import files
import numpy as np
from tensorflow.keras.preprocessing import image
import zipfile
import os

uploaded = files.upload()

# Assuming the user uploads a zip file containing images
for filename in uploaded.keys():
    if filename.endswith('.zip'):
        zip_path = f"/content/{filename}"
        extract_dir = "/content/inference_images"
        os.makedirs(extract_dir, exist_ok=True)

        with zipfile.ZipFile(zip_path, 'r') as zip_ref:
            zip_ref.extractall(extract_dir)

        print(f"Extracted images to {extract_dir}")

        # Process images in the extracted directory
        for root, _, files_in_dir in os.walk(extract_dir):
            for fname in files_in_dir:
                if fname.lower().endswith(('jpg', 'jpeg', 'png')):
                    img_path = os.path.join(root, fname)
                    try:
                        img = image.load_img(img_path, target_size=(224, 224))
                        x = image.img_to_array(img)
                        x = np.expand_dims(x, axis=0)
                        x = preprocess_input(x)   # remember we used VGG16 preprocessing
                        preds = model.predict(x)[0]
                        top3_idx = preds.argsort()[-3:][::-1]
                        print(f"\nTop-3 predictions for {img_path}:")
                        for i in top3_idx:
                            print(f"{class_names[i]} — {preds[i]*100:.2f}%")
                    except Exception as e:
                        print(f"Could not process image {img_path}: {e}")

    elif filename.lower().endswith(('jpg', 'jpeg', 'png')):
        # Process single image file
        img_path = f"/content/{filename}"
        try:
            img = image.load_img(img_path, target_size=(224, 224))
            x = image.img_to_array(img)
            x = np.expand_dims(x, axis=0)
            x = preprocess_input(x)   # remember we used VGG16 preprocessing
            preds = model.predict(x)[0]
            top3_idx = preds.argsort()[-3:][::-1]
            print(f"\nTop-3 predictions for {img_path}:")
            for i in top3_idx:
                print(f"{class_names[i]} — {preds[i]*100:.2f}%")
        except Exception as e:
             print(f"Could not process image {img_path}: {e}")
    else:
        print(f"Skipping unsupported file: {filename}")
```

```python
# Cell 14 — Classification report for test dataset
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt

# 1️⃣ Load test dataset
test_dir = '/content/inference_images/test'   # change path if needed
test_ds = tf.keras.preprocessing.image_dataset_from_directory(
    test_dir,
    image_size=(224, 224),
    batch_size=32,
    shuffle=False   # do NOT shuffle so labels align with predictions
)

# Store class names before prefetching
test_class_names = test_ds.class_names

# 2️⃣ Preprocess using same VGG16 preprocessing
from tensorflow.keras.applications.vgg16 import preprocess_input
AUTOTUNE = tf.data.AUTOTUNE
test_ds = test_ds.map(lambda x, y: (preprocess_input(x), y), num_parallel_calls=AUTOTUNE)
test_ds = test_ds.prefetch(buffer_size=AUTOTUNE)

# 3️⃣ Predict on the test set
y_true = np.array([])
y_pred = np.array([])

for images, labels in test_ds:
    preds = model.predict(images)
    y_true = np.concatenate([y_true, labels.numpy()])
    y_pred = np.concatenate([y_pred, np.argmax(preds, axis=1)])

# 4️⃣ Generate the classification report
print("Classification Report for Test Dataset:")
print(classification_report(y_true, y_pred, target_names=test_class_names))

# 5️⃣ Optional — visualize confusion matrix
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(10,8))
plt.imshow(cm, interpolation='nearest', cmap=plt.cm.Blues)
plt.title('Confusion Matrix (Test Set)')
plt.colorbar()
tick_marks = np.arange(len(test_class_names))
plt.xticks(tick_marks, test_class_names, rotation=90, fontsize=6)
plt.yticks(tick_marks, test_class_names, fontsize=6)
plt.tight_layout()
plt.ylabel('True label')
plt.xlabel('Predicted label')
plt.show()
```
